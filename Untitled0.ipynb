```python
emp_no=int(input("enter the employee numbeer:"))
emp_name=input("enter the name:")
emp_dep=input("enter the department")
emp_design=input("enter the designation")
emp_sal=float(input("enter the salary:"))
bsal=float(input("enter the basic salary:"))
if bsal>=25000:
  hra=bsal*(10/100)
  da=bsal*(15/100)
  ta=bsal*(10/100)
  lic=bsal*(20/100)
  pf=bsal*(30/100)
  it=bsal*(20/100)
elif bsal>=25000 and bsal<=50000:
  hra=bsal*(17/100)
  da=bsal*(20/100)
  ta=bsal*(15/100)
  lic=bsal*(25/100)
  pf=bsal*(38/100)
  it=bsal*(24/100)
else:
  hra=bsal*(15/100)
  da=bsal*(10/100)
  ta=bsal*(25/100)
  lic=bsal*(30/100)
  pf=bsal*(20/100)
  it=bsal*(300/100)

gross_salary=bsal+hra+da+ta
net_sal=bsal+lic+pf+it
print("-----------------------------------------------------------------------")
print("|                      SRM INSTITUTE MADRAS                           |")
print("|                       EMPLOYEE PAYSLIP                              |")
print("-----------------------------------------------------------------------")
print("ENTER EMPLOYEE NUMBER:                        ",emp_no)
print("ENTER EMPLOYEE NAME:                          ",emp_name)
print("ENTER DEPARTMENT:                             ",emp_dep)
print("ENTER EMPLOYEE DESIGNATION:                   ",emp_design)
print("ENTER EMPLOYEE SALARY:                        ",emp_sal)
print("ENTER EMPLOYEE BASIC SALARY:                  ",bsal)
print("HOUSE RENT ALLOWANCE:                         ",hra)
print("TRAVEL ALLOWANCE:                             ",ta)
print("DEARNESS ALLOWANCE:                           ",da)
print("LIFE INSURANCE CORPORATION:                   ",lic)
print("PROVIDENT FUND:                               ",pf)
print("INCOME TAX:                                   ",it)
print("-----------------------------------------------------------------------")
print("gross salary:",gross_salary)
print("net salary:",net_sal)
print("-----------------------------------------------------------------------")
```

```python
print("1.ADDITION")
print("2.SUBRACTION")
print("3.MULTIPLICATION")
print("4.DIVISION")
a=float(input("enter the 1st number:"))
b=float(input("enter the 2nd number:"))
ch=int(input("enter your choice:"))
if ch==1:
  print("ADDITION:",a+b)
elif ch==2:
  print("SUBRACTION",a-b)
elif ch==3:
  print("MUltiplication:",a*b)
elif ch==4:
  print("DIVISION:",a/b)
else:
  print("Invalid Choice")
```
