```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            low = mid + 1

        else:
            high = mid - 1

    return -1


arr = [10, 20, 30, 40, 50, 60, 70]

target = 50

result = binary_search(arr, target)

print(result)
```

```python
numbers = list(map(int,input("Enter the sorted list numbers: ").split()))
n = len(numbers)
for i in range(n):
  for j in range(0, n - i - 1):
    if numbers[j] > numbers[j + 1]:
      numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
print("Sorted list:", numbers)
```

```python
numbers = list(map(int,input("Enter the list numbers: ").split()))
n = len(numbers)
for i in range(n):
  mid_index = 1
  for j in range(i + 1, n):
    if numbers[j] < numbers[mid_index]:
      mid_index = j
  numbers[i], numbers[mid_index] = numbers[mid_index], numbers[i]
print("Sorted list:", numbers)
```

```python
numbers = list(map(int, input("Enter the list numbers: ").split()))
for i in range(1, len(numbers)):
  key = numbers[i]
  j = i - 1
  while j >= 0 and numbers[j] > key:
    numbers[j + 1] = numbers[j]
    j -= 1
  numbers[j + 1] = key
print("Sorted list:", numbers)

```
